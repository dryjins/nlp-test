# Semantic Mass Spectrometer & Lorentz Deflection
이 노트북은 LLM의 추론 과정을 의미론적 질량 분석기(Semantic Mass Spectrometer)로 모델링하여, 쿼리 토큰이 뇌터 기둥(Noether Pillar)이 형성하는 의미장 속을 통과할 때 발생하는 **로렌츠 부스트(Lorentz Boost)** 궤적을 시뮬레이션합니다.

로렌츠 인자 $\gamma$를 계산하여 해당 토큰이 안정적인 궤적 굴절(Stable Deflection)을 일으키는지, 아니면 빛의 속도를 초과하여 인과율을 붕괴시키는 **타키온 붕괴(Tachyonic Collapse = 환각)**를 일으키는지 증명합니다.

```python
import numpy as np
import pandas as pd

class MINALorentzEngine:
    def __init__(self, c_semantic=0.85):
        self.c = c_semantic
        
    def calculate_lorentz_factor(self, h, b):
        """
        속도 비율(beta)과 로렌츠 인자(gamma)를 계산합니다.
        beta >= 1 이면 빛의 속도 c를 초과한 타키온 상태이므로 gamma는 허수(NaN)가 됩니다.
        """
        if h == 0: return float('inf'), float('nan')
        beta = b / (self.c * h)
        if beta >= 1.0: return beta, float('nan')
        gamma = 1.0 / np.sqrt(1 - beta**2)
        return beta, gamma
        
    def shoot_query(self, keys):
        """
        주어진 Key 토큰들 사이로 Query를 쏘아 궤적의 굴절 상태를 평가합니다.
        """
        results = []
        for key in keys:
            k_name, k_h, k_b = key['name'], key['h'], key['b']
            
            # 민코프스키 계량에 따른 의미론적 질량 (M)
            mass_sq = k_b**2 - (self.c * k_h)**2
            mass = np.sign(mass_sq) * np.sqrt(np.abs(mass_sq))
            
            # 로렌츠 동역학 (Lorentz Dynamics)
            beta, gamma = self.calculate_lorentz_factor(k_h, k_b)
            
            # 쌍곡선 궤적 굴절각 (Rapidity, phi)
            if beta < 1.0:
                rapidity = np.arctanh(beta)
                status = "Stable Deflection"
            else:
                rapidity = float('nan')
                status = "🚨 Tachyonic Collapse (Hallucination)"
                
            results.append({
                'Token': k_name,
                'Mass (M)': round(mass, 4),
                'Velocity (β)': round(beta, 4),
                'Lorentz Factor (γ)': round(gamma, 4) if not np.isnan(gamma) else 'Complex (NaN)',
                'Deflection (φ)': round(rapidity, 4) if not np.isnan(rapidity) else 'Undefined',
                'Status': status
            })
        return results
```

```python
# Month Pillar 실험에서 도출된 H(소속감)와 B(구체성) 합성 데이터
keys = [
    {'name': 'january', 'h': 4.0, 'b': 3.3},   # 광자(Photon)에 가까운 상태 (M ~ 0)
    {'name': 'apple',   'h': 2.0, 'b': 1.0},   # 안정적 질량 입자 (구체성이 낮음)
    {'name': 'may',     'h': 3.5, 'b': 4.2},   # 다의어 (Polysemic, 무거운 질량)
    {'name': 'plans',   'h': 1.5, 'b': 5.5},   # 중력 사보추어 (Extreme B, 환각 유발자)
]

# 정보 전파 속도 c = 0.85 로 엔진 초기화
engine = MINALorentzEngine(c_semantic=0.85)
results = engine.shoot_query(keys)

# Pandas DataFrame으로 깔끔하게 출력
df = pd.DataFrame(results)
display(df)
```
