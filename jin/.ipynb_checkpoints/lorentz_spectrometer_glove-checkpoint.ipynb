# Semantic Mass Spectrometer — True Lorentz Deflection Experiment
## 실제 GloVe 임베딩 기반 어텐션 전후 질량 변화 측정

이 노트북은 `january` 쿼리 토큰이 다양한 컨텍스트(안전/사보추어/혼합) 속을 어텐션으로 통과할 때,
**민코프스키 질량(M)과 속도(β)가 어텐션 전후로 얼마나 변화하는지** 측정합니다.

$$M^2 = \|\mathbf{B}\|^2 - c^2 H^2, \quad \beta = \frac{\|B\|}{c \cdot H}$$

- $\beta < 1$: 안정적 궤적 (Stable Deflection)
- $\beta \geq 1$: 타키온 붕괴 (Tachyonic Collapse = Hallucination)

```python
import numpy as np
import gensim.downloader as api

print('Loading GloVe 100d...')
glove = api.load('glove-wiki-gigaword-100')
embeddings = glove.vectors
words_list = list(glove.key_to_index.keys())
w2i = {w: i for i, w in enumerate(words_list)}
print(f'Loaded: {embeddings.shape}')
```

```python
# MINA Core Engine
class MINAEngine:
    def __init__(self, embeddings, word_to_idx, c_semantic=0.85):
        self.embs = embeddings
        self.w2i = word_to_idx
        self.c = c_semantic
        self.pillar = None

    def induce_pillar(self, stable_words):
        indices = [self.w2i[w] for w in stable_words if w in self.w2i]
        vecs = self.embs[indices]
        p = vecs.mean(axis=0)
        self.pillar = p / np.linalg.norm(p)
        print(f'Pillar induced from: {[w for w in stable_words if w in self.w2i]}')

    def decompose(self, vec):
        h = np.dot(vec, self.pillar)
        b_vec = vec - h * self.pillar
        b = np.linalg.norm(b_vec)
        return h, b

    def mass(self, vec):
        h, b = self.decompose(vec)
        mass_sq = b**2 - (self.c * h)**2
        return np.sign(mass_sq) * np.sqrt(np.abs(mass_sq)), h, b


# Semantic Mass Spectrometer
class SemanticMassSpectrometer:
    def __init__(self, mina_engine):
        self.mina = mina_engine

    def attention(self, q_vec, key_vecs, temperature=1.0):
        d = q_vec.shape[0]
        scores = np.array([np.dot(q_vec, k) for k in key_vecs])
        scores = scores / (np.sqrt(d) * temperature)
        scores -= scores.max()
        weights = np.exp(scores)
        weights /= weights.sum()
        output = sum(w * v for w, v in zip(weights, key_vecs))
        return output, weights

    def shoot(self, query_word, context_words):
        if query_word not in self.mina.w2i:
            print(f'Word not found: {query_word}')
            return None

        q_vec = self.mina.embs[self.mina.w2i[query_word]]
        valid_keys = [w for w in context_words if w in self.mina.w2i]
        key_vecs = [self.mina.embs[self.mina.w2i[w]] for w in valid_keys]

        m_before, h_before, b_before = self.mina.mass(q_vec)
        beta_before = b_before / (self.mina.c * h_before) if h_before > 0 else float('inf')

        o_vec, attn_weights = self.attention(q_vec, key_vecs)

        m_after, h_after, b_after = self.mina.mass(o_vec)
        beta_after = b_after / (self.mina.c * h_after) if h_after > 0 else float('inf')

        status = '🚨 Tachyonic Collapse (Hallucination)' if beta_after >= 1.0 else '✅ Stable Deflection'

        return {
            'query': query_word,
            'context': valid_keys,
            'attn_weights': dict(zip(valid_keys, np.round(attn_weights, 4))),
            'BEFORE': {'M': m_before, 'H': h_before, 'B': b_before, 'β': beta_before},
            'AFTER':  {'M': m_after,  'H': h_after,  'B': b_after,  'β': beta_after},
            'ΔM': m_after - m_before,
            'Δβ': beta_after - beta_before,
            'Status': status
        }

    def print_result(self, res):
        if not res:
            return
        print('\n' + '='*70)
        print(f"  SHOOT: '{res['query']}' → Context: {res['context']}")
        print('='*70)
        print(f"  Attention Weights  : {res['attn_weights']}")
        print(f"  {'':25} {'Mass (M)':>8} {'Height (H)':>10} {'Base (B)':>8} {'Beta (β)':>9}")
        print(f"  {'BEFORE (query)':25} {res['BEFORE']['M']:8.4f} {res['BEFORE']['H']:10.4f} {res['BEFORE']['B']:8.4f} {res['BEFORE']['β']:9.4f}")
        print(f"  {'AFTER  (attn out)':25} {res['AFTER']['M']:8.4f} {res['AFTER']['H']:10.4f} {res['AFTER']['B']:8.4f} {res['AFTER']['β']:9.4f}")
        print(f"  {'DELTA':25} {res['ΔM']:8.4f} {'':10} {'':8} {res['Δβ']:9.4f}")
        print(f"\n  Status: {res['Status']}")
        print('='*70)
```

```python
# Pillar 유도 (안정적 월 단어 기준)
mina = MINAEngine(embeddings, w2i, c_semantic=0.85)
stable_months = ['january', 'february', 'march', 'april', 'august', 'september']
mina.induce_pillar(stable_months)

spectrometer = SemanticMassSpectrometer(mina)
```

```python
# Experiment 1: 안전한 컨텍스트 (Months only)
res1 = spectrometer.shoot('january', ['february', 'march', 'april', 'june'])
spectrometer.print_result(res1)
```

```python
# Experiment 2: 사보추어 주입 (plans, started)
# 핵심 관찰: 소수의 사보추어가 전체 궤적을 붕괴시키는가?
res2 = spectrometer.shoot('january', ['february', 'plans', 'started'])
spectrometer.print_result(res2)
```

```python
# Experiment 3: 다의어 혼합 (may, apple, june)
res3 = spectrometer.shoot('january', ['may', 'apple', 'june'])
spectrometer.print_result(res3)
```

```python
# Experiment 4: 사보추어 비율 스캔
# 사보추어의 비중이 얼마나 되어야 붕괴가 발생하는가?
safe_tokens  = ['february', 'march', 'april']
saboteur     = 'plans'

print('\n' + '='*70)
print(f"{'SABOTEUR RATIO SCAN':^70}")
print('='*70)
print(f"{'Context':<35} {'β_after':>8} {'ΔM':>8} {'Status'}")
print('-'*70)

for n_sab in range(0, 5):
    context = safe_tokens + [saboteur] * n_sab
    res = spectrometer.shoot('january', context)
    if res:
        ctx_str = f"{safe_tokens} + {n_sab}x'{saboteur}'"
        status_icon = '🚨' if res['AFTER']['β'] >= 1.0 else '✅'
        print(f"{ctx_str:<35} {res['AFTER']['β']:8.4f} {res['ΔM']:8.4f} {status_icon}")
print('='*70)
```
